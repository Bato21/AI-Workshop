```python
import pandas as pd
import matplotlib.pyplot as plt

curves = pd.DataFrame({
    "training_rows": [50, 100, 200, 400, 800],
    "A_train_accuracy": [0.61, 0.62, 0.63, 0.64, 0.65],
    "A_validation_accuracy": [0.56, 0.58, 0.60, 0.61, 0.62],
    "B_train_accuracy": [0.99, 0.98, 0.96, 0.95, 0.94],
    "B_validation_accuracy": [0.53, 0.58, 0.62, 0.65, 0.66],
})
display(curves)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
for ax, label in zip(axes, ["A", "B"]):
    ax.plot(curves.training_rows, curves[f"{label}_train_accuracy"], marker="o", label="Training")
    ax.plot(curves.training_rows, curves[f"{label}_validation_accuracy"], marker="o", label="Validation")
    ax.set_title(f"Fictional case {label}")
    ax.set_xlabel("Rows used for training")
    ax.set_ylabel("Accuracy")
    ax.set_ylim(0.45, 1.0)
    ax.grid(alpha=0.25)
    ax.legend()
plt.tight_layout()
plt.show()
```
